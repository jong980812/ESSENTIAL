## Text file ( filename, label ) convert to (filename, class )
### label -> 1,2,3,..., 400
### class name -> playing basketball.....

```python

class_dict = {}
class_label_mathing = '/data/jong980812/project/videoCIL/kinetics400_label.txt'
filename_label= '/data/jong980812/project/videoCIL/kinetics400_test_list_videos.txt'
filename_class = filename_label.split('.')[0]+'(filname_class)'+'.txt'
with open(class_label_mathing, 'r') as label_file:
    for line in label_file:
        line = line.strip().split(',')
        label = int(line[0])
        class_name = line[1]
        class_dict[label] = class_name

with open(filename_label, 'r') as input_file:
    with open('kinetics400_test_list_videos(filename_class).txt', 'w') as output_file: 
        for line in input_file:
            line = line.strip().split()
            filename = line[0]
            label = int(line[1])
            class_name = class_dict.get(label, 'Unknown')  # 클래스 정보가 없을 경우 'Unknown'으로 처리
            output_file.write(f"{filename},{class_name}\n")

print("새로운 텍스트 파일이 생성되었습니다.")
```

# Validation

```python
import pickle

# Open the .pkl file
with open("/data/datasets/Activitynet/Kinetics400_data_tasks_20.pkl", "rb") as file:
    task_20 = pickle.load(file)
# Print the loaded data

# 기존 .pkl 파일 열기
data = task_20

# 새로운 데이터 구조 생성 (기존 데이터를 복제)
new_data = data.copy()

# txt 파일에서 파일 이름과 클래스 이름 읽어오기


#새로운 pkl생성 위해서 딕셔너리 초기화
for split in new_data:
    for i,item in enumerate(new_data[split]):
        # 파일 이름을 해당 클래스 이름의 리스트에 추가
        for each_class in item:
            new_data[split][i][each_class]=[]
            

txt_file_path = "/data/jong980812/project/videoCIL/kinetics400_train_list_videos(filename_class).txt"
total_num=0      
with open(txt_file_path, "r") as txt_file:
    for line in txt_file:
        filename, class_name = line.strip().split(",")
        for i,item in enumerate(new_data['train']):
            for classs in item:
                if classs==class_name:
                    total_num+=1
                    new_data['train'][i][classs].append(filename) 
print(total_num)
                    
txt_file_path = "/data/jong980812/project/videoCIL/kinetics400_val_list_videos(filename_class).txt"
total_num=0      
with open(txt_file_path, "r") as txt_file:
    for line in txt_file:
        filename, class_name = line.strip().split(",")
        for i,item in enumerate(new_data['val']):
            for classs in item:
                if classs==class_name:
                    total_num+=1
                    new_data['val'][i][classs].append(filename) 
print(total_num)
txt_file_path = "/data/jong980812/project/videoCIL/kinetics400_test_list_videos(filename_class).txt"
total_num=0                  
with open(txt_file_path, "r") as txt_file:
    for line in txt_file:
        filename, class_name = line.strip().split(",")
        for i,item in enumerate(new_data['test']):
            for classs in item:
                if classs==class_name:
                    total_num+=1
                    new_data['test'][i][classs].append(filename) 
print(total_num)
```
