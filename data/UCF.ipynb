```python
import pickle

import pandas as pd

# 딕셔너리로 저장할 변수를 생성
result_dict = {}

# 텍스트 파일을 열어서 각 줄을 읽어온다
with open('/data/bkh178/framework/VideoMAE_cil/data/ucf_classInd.txt', 'r') as file:
    lines = file.readlines()
    for line in lines:
        # 각 줄을 공백을 기준으로 나누어 key와 value로 분리
        key, value = line.strip().split(' ', 1)
        # key를 정수로 변환하여 딕셔너리에 저장
        # set start inx zero
        result_dict[value] = int(key)-1

# 결과 출력
print(result_dict)



with open('/data/bkh178/framework/vCLIMB_Benchmark/data/UCF101_data_20tasks.pkl', 'rb') as file:
    data = pickle.load(file)

cil_names = []
for i in data['train']:
    cil_names.append(list(i.keys()))
_cil_names = set(item for sublist in cil_names for item in sublist)

assert set(_cil_names) == set(result_dict.keys())
assert len(set(_cil_names)) == 101
```

```python
converted_list = [[result_dict[label] for label in sublist] for sublist in cil_names]
flattened_list = [item for sublist in converted_list for item in sublist]

# 리스트를 집합으로 변환
unique_set = set(flattened_list)
assert 101 == len(unique_set)
converted_list
```

```python
my_list = [[] for _ in range(len(converted_list))]
sum = 0
for n,i in enumerate(data['val']):
    for label, lst in i.items():
        sum += len(lst)
        for video in lst:
            my_list[n].append(f'{label}/{video}.avi {result_dict[label]}')
            
with open('ucf_20_task/val.pkl', 'wb') as f:
    pickle.dump(my_list, f)

print(len(set(item for sublist in my_list for item in sublist)))
```
