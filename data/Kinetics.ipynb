```python
import pickle

import pandas as pd

data = pd.read_csv('/data/bkh178/framework/vCLIMB_Benchmark/data/kinetics_label.csv', index_col='id')
result_dict = data['name'].to_dict()
result_dict = {v: k for k, v in result_dict.items()}

print(result_dict)


with open('/data/bkh178/framework/vCLIMB_Benchmark/data/Kinetics400_data_tasks_20.pkl', 'rb') as file:
    data = pickle.load(file)

cil_names = []
for i in data['train']:
    cil_names.append(list(i.keys()))
cil_names[0]
```

```python
converted_list = [[result_dict[label] for label in sublist] for sublist in cil_names]
flattened_list = [item for sublist in converted_list for item in sublist]

# 리스트를 집합으로 변환
unique_set = set(flattened_list)
assert 400 == len(unique_set)
converted_list
if 0 in unique_set:
    print("yes")
```

```python
import csv

my_list = [[] for _ in range(len(converted_list))]

with open('/data/bkh178/framework/VideoMAE/data/train.csv', 'r') as f:
    reader = csv.reader(f)
    for i, labels in enumerate(converted_list):
        f.seek(0)  # 파일 포인터를 파일의 시작으로 이동
        for row in reader:
            path, label = row[0].split(' ')
            label = int(label)
            if label in labels:
                my_list[i].append(row[0])
total_elements = sum(len(inner_list) for inner_list in my_list)
print(total_elements)



```
